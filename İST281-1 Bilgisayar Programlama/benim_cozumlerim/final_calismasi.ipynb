```python
# Soru 9 
a = 1
b = 4
c = -12
delta = b**2-4*a*c
x_1 = (-b-(delta**(1/2)))/(2*a)
x_2 = (-b+(delta**(1/2)))/(2*a)

print("Kökler sırasıyla: "+str(x_1)+" ve "+str(x_2)+" dir.")
```
