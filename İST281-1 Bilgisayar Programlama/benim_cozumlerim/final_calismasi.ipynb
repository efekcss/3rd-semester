### Örnekler

```python
# Soru 1
sayi = int(input("Sayi : "))

if sayi%2==0:
    print("Çift.")
else:
    print("Tek.")
```

```python
# Ek 1
sayi = int(input("Sayi : "))

if sayi%2==0:
    if sayi%4==0:
        print("4'e tam bölünür.")
    else:
        print("4'e tam bölünmez.")
else:
    print("Tek.")
```

```python
# Ek 2
sayi1 = int(input("1.sayi :"))
sayi2 = int(input("2.sayi :"))

if sayi1%sayi2 == 0:
    print("Tam bölünüyor.")
else:
    print("Tam bölünmüyor.")
```

```python
# Soru 2
ulke = "1" #US
tutar = int(input("Alışveriş Tutarı = "))
if ulke == "1":
    if tutar>150:
        print("Ucretsiz")
    elif 150>=tutar>100:
        print("1 TL")
    elif 100>=tutar>50:
        print("3 TL")
    elif tutar <= 50:
        print("5 TL")
else: # AU
    if tutar>=50:
        print("Ucretsiz.")
    else:
        print("10 TL")
```

```python
# Soru 3
s1 = int(input("S1:"))
s2 = int(input("S2:"))
s3 = int(input("S3:"))

if s1>s2:
    if s2>s3:
        print(f"1.Sayi olan {s1} en büyüktür.")
    elif s3>s1:
        print(f"3.Sayi olan {s3} en büyüktür.")
elif s2>s3:
    print(f"2.sayi olan {s2} en büyüktür.")
else:
    print(f"3.sayi olan {s3} en büyüktür.")
```

```python
# Soru 3
# b)
s1 = int(input("S1:"))
s2 = int(input("S2:"))
s3 = int(input("S3:"))

if s1>s2 and s1>s3:
    print(f"1.sayi olan {s1} en buyuktur.")
elif s2>s1 and s2>s3:
    print(f"2.Sayi olan {s2} en buyuktur.")
else:
    print(f"3.Sayi olan {s3} en buyuktur.")
```
