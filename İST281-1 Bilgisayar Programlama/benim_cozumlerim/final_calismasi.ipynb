# Ders_8_Fonksiyonlar-1.pdf
### Fonksiyonlar örnek sorular

```python
# return mantığı önemli onun için bir örnek göstericem. 
# return değerin tutulmasına. döndürülebilmesine yarar yoksa 
# return kullanılmazsa sadece sonuç "gösterir"

def topla1(a,b):
    c = a+b

toplam1 = topla1(5,9)
print(toplam1)

def topla2(a,b):
    c = a+b
    return c

toplam2 = topla2(5,9)
print(toplam2)

# çıktı da görüldüğü gibi return kullanmadığımız topla1 fonksiyonu
# None değeri döndürdü ama topla2 olan return kullandığımız değeri bize
# döndürebildi
```

```python
# Soru 1
def tek_cift(n):
    if n%2==0:
        return int(n/2)
    else:
        return 3*n+1

print(tek_cift(4))
print(tek_cift(5))        
```

```python
# Soru 2
def yeni_tek_cift(n):
    while n != 1:
        if n%2 == 0 :
            n /= 2
            print(n) # ara işlemleri görebilmek için  
        else:
            n = 3*n+1
            print(n)
    return n
print(yeni_tek_cift(6))
```

```python
# Soru 4


def clean_list(liste):
    yeni_liste = []
    for i in liste:
        if i not in yeni_liste:
            yeni_liste.append(i)
    return yeni_liste
a = [3,6,4,4,3,2,7,8]
print(clean_list(a))
```

```python
# Soru 4

l1 = [1,2,3]
l2 = [3,1,4]

liste = []

for i in l1:
    for j in l2:
        if i != j:
            liste.append([i,j])
print(liste)
```

```python
# Soru 5 
liste = []
for i in l1:
    for j in l2:
        if i != j:
            liste.extend([i,j])
print(liste)
```

```python
# Soru 6

n = int(input("n : "))
liste = []
for i in range(1,n):
    liste.append(i)
tek_liste = []
cift_liste = []
for j in liste:
    if j%2==0:
        cift_liste.append(j)
    else:
        tek_liste.append(j)
print(tek_liste)
print(cift_liste)
```

```python
# Soru 7 
n = int(input("n : "))
liste = []
for i in range(n):
    liste.append(int(input("Sayi giriniz : ")))
if n % 2 != 0:
    liste.append(min(liste))
    n += 1
devrik = liste[::-1]
for j in range(n): # range(len(liste)) kullanımı n düzeltmesine gerek kalmadan daha doğru bir karar olurmuş du.
    print(f"{liste[j]} + {devrik[j]} = {liste[j] + devrik[j]}")
```
