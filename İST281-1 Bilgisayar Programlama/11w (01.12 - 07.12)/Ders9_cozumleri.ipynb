# Ders 9 Liste Metotları (List Methods) Örnek Soruları Çözümleri

```python
# Soru 1 

a = []
for i in range(10):
    a.append(i*i)

print(a)
```

```python
# Soru 2 

a = [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

for i in a:
    if i % 2 != 0:
        a.remove(i)

print(a)
```

```python
# Soru 3

n = int(input("n = "))

isimler = []

for i in range(n):
    isimler.append(input(f"İsim #{i+1} = "))
for j in isimler:
    print(j,end="\n")
```
