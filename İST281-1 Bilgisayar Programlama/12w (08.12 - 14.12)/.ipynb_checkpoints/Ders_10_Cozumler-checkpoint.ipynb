# Lesson #10 List Comprehensions (Liste Üreteçleri)

### Liste üreteçleri 
Liste üretmek için kısa yol sağlar(kısa kod).Amaç tekrarlanabilen (iterable) işlemler ile gerekirse koşul ekleyerek yeni listeler oluşturmaktır.

```python
kare = []

for x in range(1,7):
    kare.append(x**2)
print(kare)

# bunun yerine daha kısa olan bir "list comprehension" kullanabiliriz : 

kare_compred = [x**2 for x in range(1,7)]

print(kare_compred)
```

```python
# list comprehensions ile koşullu kullanıma bir örnek :
vector = [-4,-2,0,2,4]
vector_pos = [x for x in vector if x >= 0]

print(vector_pos)

```

```python
# Liste öğelerinde "method" çağırılabilir.

taze_meyve = ["muz","kizilcik","carkifelek"]

t = [silah.capitalize() for silah in taze_meyve]

print(t)
```

```python
# Alt liste oluşturulabilir.

matris = [[x,x**2] for x in range(6)]
print(matris) 
```

```python
# İki for deyimi ile alt listeler, düz listeye dönüştürülebilir.

vector = [[1,2,3],[4,5,6],[7,8,9]]

liste = [sayi for oge in vector for sayi in oge]

print(liste)
```
